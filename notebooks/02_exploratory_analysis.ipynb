# Exploratory analysis
Explore monthly cash flow, category spending and transaction distribution.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/sample/transactions.csv', parse_dates=['date'])
monthly = df.assign(month=df.date.dt.to_period('M')).groupby(['month','type']).amount.sum().unstack(fill_value=0)
monthly.plot(kind='bar', title='Monthly income and expenses')
plt.show()
```
