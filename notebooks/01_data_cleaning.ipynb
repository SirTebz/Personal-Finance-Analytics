# Data cleaning
Load the synthetic CSV/export, inspect types, missing values and duplicate rows before analysis.

```python
import pandas as pd
df = pd.read_csv('../data/sample/transactions.csv', parse_dates=['date'])
df.info()
df = df.drop_duplicates().dropna(subset=['date','type','amount'])
df['amount'] = pd.to_numeric(df['amount'], errors='coerce')
cleaned = df[df['amount'].gt(0)]
```
