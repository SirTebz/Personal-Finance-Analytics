# Financial insights
Calculate reproducible descriptive insights; avoid causal or advisory claims.

```python
import pandas as pd
df = pd.read_csv('../data/sample/transactions.csv')
income = df.loc[df.type.eq('Income'), 'amount'].sum()
expenses = df.loc[df.type.eq('Expense'), 'amount'].sum()
savings_rate = (income-expenses)/income*100 if income else 0
top_category = df.loc[df.type.eq('Expense')].groupby('category').amount.sum().idxmax()
print(f'Savings rate: {savings_rate:.1f}%; highest spending category: {top_category}.')
```
